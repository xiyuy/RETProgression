```python
import pandas as pd

# Load the CSV
csv_path = '/projects/retprogression/clean_gradable_dr_test.csv'
output_csv = '/projects/retprogression/clean_ungradable_dr_test_with_links.csv'
bucket_name = "ai_research_images_store"

# Load DataFrame
df = pd.read_csv(csv_path)

# Filter for gradable_binary_DR == 0
ungradable_df = df[df['gradable_binary_DR'] == 0].copy()

# Determine the image ID column
id_col = 'ID' if 'ID' in ungradable_df.columns else 'ID' if 'ID' in ungradable_df.columns else None

if id_col is None:
    raise ValueError("No ID column found. Expected one named 'image_id' or 'id'.")

# Create the clickable public URL
ungradable_df['public_url'] = ungradable_df[id_col].astype(str).apply(
    lambda x: f"https://storage.cloud.google.com/{bucket_name}/{x}.jpg?authuser=0"
)

# Save output
output_df = ungradable_df[[id_col, 'public_url']]
output_df.to_csv(output_csv, index=False)

print(f"✅ Clickable image links saved to: {output_csv}")

```
